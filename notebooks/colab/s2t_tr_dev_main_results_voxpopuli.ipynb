# VoxPopuli Main Results — PARKED

**Status:** Not part of the current submission.

Across four config versions (`v1` -> `v4`) the proposed hierarchical
transformer router collapses to picking HuBERT 100% of the time on
VoxPopuli, despite Whisper being the per-clip oracle on ~46% of
clips. The configs are kept under
[`configs/experiments/deprecated/`](../../configs/experiments/deprecated/)
with a detailed timeline in
[`configs/experiments/deprecated/README.md`](../../configs/experiments/deprecated/README.md)
and a one-paragraph blocker in
[`PROJECT_STATE.md`](../../PROJECT_STATE.md) §5.

Reviving this deliverable would likely require one of:
1. Replacing Whisper-base with Whisper-large.
2. Per-clip outlier clipping in the soft-CE loss.
3. A different routing target (e.g. weighted ROVER as the oracle).

None of these are scoped for the current paper. **Do not run this
notebook expecting a working result.**

If you want to resurrect the failure to attempt a fix, the historical
make targets are still available:

```
make run_main_results_voxpopuli      # the v1 attempt
make run_main_results_voxpopuli_v2   # +loss-fix attempt
make run_main_results_voxpopuli_v3   # +risk-taking soft-CE attempt
make run_main_results_voxpopuli_v4   # +dynamic class-balancing attempt
```

but each of those is wired to a `configs/experiments/deprecated/<...>.yaml`
and PROJECT_STATE.md should be updated before re-running.
